## Import data

```python
with open("data/mlops_raw.txt", "r") as file:
    mlops_text = file.read()

mlops_text[:150]
```

## Data models

```python
from pydantic import BaseModel, Field

class education(BaseModel):
    education_name:str = Field(description="Make sure to find this information in the text")
    program_description:str = Field(description="A 3-5 sentences summary about the program")
    carrer_options:list[str]
    courses: list[str]
    length:float = Field(description="Total study length of the program")
```

## Create agent

```python
from pydantic_ai import Agent

education_agent = Agent(
    "openrouter:openai/gpt-oss-120b:free",
    system_prompt="""
    You are an expert in swedish education focused on Yrkeshögskola (YH).
    Based on a text avout an education, you will extract out relevant information.   

    Make sure to have the same language as the text    
""",                      
)

result = await education_agent.run(mlops_text, output_type=education)
result.output
```
