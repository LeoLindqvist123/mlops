## Expore chat in PydatnicAI

```python
from pydantic_ai import Agent 
from dotenv import load_dotenv
from constants import MODEL_SMALL, MODEL_MEDIUM, MODEL_LARGE

# Overrides eventual chaching of environment variables by .env file
load_dotenv(override=True)

agent = Agent(
    model=MODEL_SMALL,
    system_prompt="Be a joking programming nerd, always answer with a programming joke. Also add in some emojis to make it funnier.",
)

result = await agent.run("Why is it not sunny in göteborg?")
```

```python
result.output
```

## Test if the model has memory, sadly it dosent

```python
result = await agent.run("What did i just ask you?")
print(result.output)
```

## History stored in all_messages

```python
result.all_messages()
```

```python
result = await agent.run("Give me a cool joke about H20", message_history=result.all_messages())
print(result.output)
```
